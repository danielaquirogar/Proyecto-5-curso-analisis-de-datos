```python
# Función para encriptar
# Recibe texto como parámetro.
def encriptar(texto):
    abecedario = "ABCDEFGHIJKLMNÑOPQRSTUVWXYZ"
    abecedario_rotado = "NÑOPQRSTUVWXYZABCDEFGHIJKLM"
# Aquí se irá construyendo el texto encriptado letra por letra
    resultado = ""

    for caracter in texto:
# La siguiente línea convierte el caracter a mayúscula y verifica si está dentro del abecedario
        if caracter.upper() in abecedario:
# Después se busca la posición de la letra original Y se toma la letra en esa misma posición del abecedario rotado.
# Esa será la letra encriptada.
            indice = abecedario.index(caracter.upper())
            nuevo_caracter = abecedario_rotado[indice]

            # Para mantener mayúsculas o minúsculas
            if caracter.islower():
                nuevo_caracter = nuevo_caracter.lower()

            resultado += nuevo_caracter
        else:
            # Para que los caracteres no reconocidos no se modifiquen
            resultado += caracter

    return resultado

# Función para desencriptar
# Recibe como parámetro un texto ya encriptado.
def desencriptar(texto_encriptado):
    abecedario = "ABCDEFGHIJKLMNÑOPQRSTUVWXYZ"
    abecedario_rotado = "NÑOPQRSTUVWXYZABCDEFGHIJKLM"
# Aquí se irá formando el texto desencriptado.
    resultado = ""

    for caracter in texto_encriptado:
# Luego se convierte el carácter a mayúscula y verifica si está en el abecedario rotado.
# Permitiendo trabajar con letras mayúsculas y minúsculas.
        if caracter.upper() in abecedario_rotado:
# Posteriormente busca la posición de la letra encriptada dentro del abecedario rotado.
# Y se toma la letra que está en la misma posición del original. Esa letra es la original antes de la encriptación.
            indice = abecedario_rotado.index(caracter.upper())
            nuevo_caracter = abecedario[indice]
# Permite mantener mayúsculas o minúsculas
            if caracter.islower():
                nuevo_caracter = nuevo_caracter.lower()
# Para después añadir el carácter desencriptado al texto final a través de la siguiente línea.
            resultado += nuevo_caracter
        else:
            resultado += caracter

    return resultado

# Pruebas del programa. Con todas las reglas y condiciones de bordes mencionadas en instrucciones.
# Por eso decido incluir: Mayúsculas y Minúsculas, Números, Signos de puntuación, Espacios y Frases completas.

def pruebas():
# def pruebas() no recibe parámetros porque los textos de prueba ya están definidos dentro de la función.
    pruebas_texto = [
        "Hola",
        "HOLA!",
        "H4L1",
        "Hola como te va",
        "Ñandú",
        "abcXYZ",
        "12345",
        "¿Hola, cómo estás?"
    ]

    for texto in pruebas_texto:
        encriptado = encriptar(texto) # Llama a función encriptar() y guarda resultado en la variable encriptado.
        desencriptado = desencriptar(encriptado) # Desencripta texto encriptado para comprobar que se obtiene el texto original.

        print("Texto original:     ", texto)
        print("Texto encriptado:   ", encriptado)
        print("Texto desencriptado:", desencriptado)
        print("-" * 40)

# Ejecutar pruebas
pruebas()
```
